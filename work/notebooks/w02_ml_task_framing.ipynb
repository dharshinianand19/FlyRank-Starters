## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row = one pseudonymized content page. I will use measurable performance and content characteristics such as impressions, CTR, engagement rate, position, content age, freshness, and word count.

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/dharshinianand19/FlyRank-Starters/main/data/raw/content_refresh_anonymized.csv"
```

```python
df = pd.read_csv(url)
```

```python
cluster_df = df[
    [
        "content_id",
        "impressions_90d",
        "ctr",
        "engagement_rate",
        "avg_position",
        "content_age_days",
        "days_since_last_update",
        "word_count"
    ]
].copy()

print(f"Rows: {len(cluster_df):,}")
print(f"Columns: {cluster_df.shape[1]}")
display(cluster_df.head())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


A fixed rule would need many separate thresholds for visibility, engagement, position, freshness, and content depth. These signals can interact in different ways across pages. Clustering can discover recurring combinations of these signals and help us describe different page archetypes without deciding the groups in advance.
