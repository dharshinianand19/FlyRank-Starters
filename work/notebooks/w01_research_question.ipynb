## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/dharshinianand19/FlyRank-Starters/main/data/raw/content_refresh_anonymized.csv"
```

```python
df = pd.read_csv(url)

print(f"Dataset: {len(df):,} rows × {df.shape[1]} columns")
print(f"Mean CTR: {df['ctr'].mean():.3f}")

print("\nMean CTR by position:")
print(df.groupby("position_tier")["ctr"].mean().round(3))

print(f"\nPages with 500+ impressions: {(df['impressions_90d'] >= 500).sum():,}")
```
